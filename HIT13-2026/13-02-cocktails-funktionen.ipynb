```python
import pandas as pd
df = pd.read_csv("cocktails.csv")
df.head()
```

```python
derString = df.ingredients.iloc[0]
df[df.name == "Black Russian"].name.str.split(" ")
df[df.name == "Black Russian"].name.str.capitalize().tolist()

# isdigit
```

```python
df.ingredients.str.split(", ")
splittedIngredients = df.ingredients.str.split(", ").explode().str.replace(" ml", "")

for i in splittedIngredients:
    for j in i:
        if j.isdigit() == True:
            i.strip(j)

# splittedIngredients
```

```python
listOfIngredients = []
for i in df.ingredients:
    ingredient = i.split(",")
    for word in ingredient:
        word = word.replace(" ml ", "")
        cleanedIngredient = ''.join(filter(lambda x: not x.isdigit(), word)).strip(".").strip()
        listOfIngredients.append(cleanedIngredient)

len(listOfIngredients), listOfIngredients
```
