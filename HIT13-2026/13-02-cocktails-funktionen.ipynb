# Aufgabe 1
Lass Dir aus der cocktails.csv eine bereinigte Zutatenliste ausgeben

```python
import pandas as pd
df = pd.read_csv("cocktails.csv")
df.head()
```

```python
derString = df.ingredients.iloc[0]
df[df.name == "Black Russian"].name.str.split(" ")
df[df.name == "Black Russian"].name.str.capitalize().tolist()

# isdigit
```

```python
df.ingredients.str.split(", ")
splittedIngredients = df.ingredients.str.split(", ").explode().str.replace(" ml", "")

for i in splittedIngredients:
    for j in i:
        if j.isdigit() == True:
            i.strip(j)

# splittedIngredients
```

```python
for i in "20 Ziegen":
    print(i.isdigit())
```

```python
for i in df.ingredients:
    zutaten = i.split(",")
    for ding in zutaten:
        ding = ding.replace(" ml ", "")
        cleaned_zutat = ''.join(filter(lambda x: not x.isdigit(), ding)).strip(".").strip()
    print(cleaned_zutat)
```
