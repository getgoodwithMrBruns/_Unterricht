```python
import os
import pandas as pd

# Ordner mit CSV-Dateien (hier aktuelles Verzeichnis)
ordner = "."

# Alle CSV-Dateien im Ordner einsammeln
klassendateien = {
    os.path.splitext(f)[0]: os.path.join(ordner, f)
    for f in os.listdir(ordner)
    if f.lower().endswith(".csv")
}

# Zielstruktur (Spalten in gewünschter Reihenfolge)
spalten = [
    "Name", "Vorname", "September", "Oktober", "November", "Dezember",
    "Januar", "1. Halbjahr", "Februar", "März", "April", "Mai", "Juni",
    "Juli", "2. Halbjahr", "Zeugnis"
]

# Mapping von CSV-Spaltennamen -> gewünschte Namen
mapping = {
    "Nachname": "Name",
    "Vorname": "Vorname"
    # ggf. weitere Zuordnungen ergänzen
}

# Neue Excel-Datei schreiben
with pd.ExcelWriter("alle_klassen.xlsx", engine="openpyxl") as writer:
    etwas_geschrieben = False

    for blattname, datei in klassendateien.items():
        try:
            print(f"Bearbeite: {blattname} ({datei})")
            df = pd.read_csv(datei)

            # Spalten umbenennen, falls sie anders heißen
            df = df.rename(columns=mapping)

            # Fehlende Spalten ergänzen
            for s in spalten:
                if s not in df.columns:
                    df[s] = ""

            # Spalten in gewünschte Reihenfolge bringen
            df_strukturiert = df[spalten]

            # In Excel-Blatt schreiben
            df_strukturiert.to_excel(writer, sheet_name=blattname, index=False)
            etwas_geschrieben = True

        except Exception as e:
            print(f"❌ Fehler bei {datei}: {e}")

    # Fallback: Wenn nichts geschrieben wurde, Dummy-Blatt einfügen
    if not etwas_geschrieben:
        pd.DataFrame({"Hinweis": ["Keine gültigen Daten gefunden"]}).to_excel(
            writer, sheet_name="Fehler", index=False
        )
```

```python
import os
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Border, Side

# Definiere einen dunkleren Rahmen
dunklerRand = Border(
    left=Side(border_style="thin", color="000000"),
    right=Side(border_style="thin", color="000000"),
    top=Side(border_style="thin", color="000000"),
    bottom=Side(border_style="thin", color="000000")
)

for sheetname in wb.sheetnames:
    ws = wb[sheetname]

    # Gitternetzlinien sichtbar lassen
    ws.sheet_view.showGridLines = True

    # Auto-Spaltenbreite
    for col in ws.columns:
        max_length = 0
        col_letter = col[0].column_letter
        for cell in col:
            if cell.value:
                max_length = max(max_length, len(str(cell.value)))
            # Dunkle Rahmenlinie anwenden
            cell.border = dunklerRand
        ws.column_dimensions[col_letter].width = max_length + 2

ordner = "."

klassendateien = {
    os.path.splitext(f)[0]: os.path.join(ordner, f)
    for f in os.listdir(ordner)
    if f.lower().endswith(".csv")
}

spalten = [
    "Name", "Vorname", "September", "Oktober", "November", "Dezember",
    "Januar", "1. Halbjahr", "Februar", "März", "April", "Mai", "Juni",
    "Juli", "2. Halbjahr", "Zeugnis"
]

mapping = {
    "Nachname": "Name",
    "Vorname": "Vorname"
}

excel_datei = "alle_klassen.xlsx"

# Schritt 1: Excel-Datei mit Pandas erzeugen
with pd.ExcelWriter(excel_datei, engine="openpyxl") as writer:
    etwas_geschrieben = False
    for blattname, datei in klassendateien.items():
        try:
            df = pd.read_csv(datei)
            df = df.rename(columns=mapping)

            for s in spalten:
                if s not in df.columns:
                    df[s] = ""

            df_strukturiert = df[spalten]
            df_strukturiert.to_excel(writer, sheet_name=blattname, index=False)
            etwas_geschrieben = True
        except Exception as e:
            print(f"❌ Fehler bei {datei}: {e}")

    if not etwas_geschrieben:
        pd.DataFrame({"Hinweis": ["Keine gültigen Daten gefunden"]}).to_excel(
            writer, sheet_name="Fehler", index=False
        )

# Schritt 2: Excel nachbearbeiten (Formatierung mit openpyxl)
wb = load_workbook(excel_datei)

for sheetname in wb.sheetnames:
    ws = wb[sheetname]

    # Gitternetzlinien aktivieren
    ws.sheet_view.showGridLines = True

    # Auto-Spaltenbreite
    for col in ws.columns:
        max_length = 0
        col_letter = col[0].column_letter  # Spaltenbuchstabe
        for cell in col:
            try:
                if cell.value:
                    max_length = max(max_length, len(str(cell.value)))
            except:
                pass
        adjusted_width = max_length + 2
        ws.column_dimensions[col_letter].width = adjusted_width

wb.save(excel_datei)
```
